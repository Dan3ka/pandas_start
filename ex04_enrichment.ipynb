```python
import pandas as pd
import numpy as np
df=pd.read_json('../data/auto.json', lines=True)
pd.options.display.float_format='{:.2f}'.format
df
```

## 2. Enrich the dataframe using a sample from that dataframe.
   - Create a sample with 200 new observations using `random_state = 21`.

- The sample should not contain new combinations of the `car number`, `make`, and `model`, so the entire dataset will be consistent in this regard.
- There are no restrictions on the `refund` and `fines` columns. You can randomly select a value from these columns and apply it to any car number.

```python
no_dupl_df=df.drop_duplicates(subset=['CarNumber', 'Make', 'Model'])
new_sample=no_dupl_df.sample(200, replace= True, random_state=21)
new_sample.loc[:,'Refund']=np.random.randint(1,3, len(new_sample))
new_sample.loc[:,'Fines']=np.random.randint(100,df['Fines'].max()+100, len(new_sample))
concat_rows=pd.concat([df, new_sample],ignore_index=True)
concat_rows
```

## 3. Enrich the `concat_rows` dataframe with a new column containing generated data.
   - Create a series named "Year" with random integers from 1980 to 2019.
   - Use `np.random.seed(21)` before generating the years.
   - Concatenate the series with the data frame and name it `fines`.

```python
np.random.seed(21)
year_series=np.random.randint(1980,2020,size= len(df))
res_year=pd.Series(year_series, name='Year')
res_year
```

```python
df_with_year=pd.concat([df,res_year], axis=1)
df_with_year.rename(columns={'Year': 'fines'}, inplace=True)
df_with_year
```

## 4. Enrich the dataframe with data from another dataframe.
   - Create a new dataframe with car numbers and their owners.
      - Get the most popular surnames in the US (you can find the file [surname.json]
      - Create a new series with the surnames. They should not contain special characters, such as commas or brackets. The count should equal the number of unique car numbers in the sample (use `random_state = 21`).
      - Create the dataframe `owners` with two columns: `CarNumber` and `SURNAME`.

```python
df_from_sur=pd.read_json('../data/surname.json')
df_from_sur.columns=df_from_sur.iloc[0]
df_from_sur=df_from_sur[1:].reset_index(drop=True)
surnames=pd.Series(df_from_sur['NAME'])
uniq_nums=df['CarNumber'].nunique()
sample_sur=surnames.sample(uniq_nums,replace=True, random_state=21)
sample_sur.reset_index(drop=True, inplace=True)
owners=pd.DataFrame({'CarNumber': df['CarNumber'].unique(),
                    'Names': sample_sur})
owners
```
